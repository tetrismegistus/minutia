```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a sample graph
G = nx.Graph()
G.add_edge("Thea", "Diana", relationship="between friends and dating")
G.add_edge("Thea", "Sasha", relationship="friends")
G.add_edge("Diana", "Owl", relationship="dating")
G.add_edge("Original", "Diana", relationship = "Aunt #2")
G.add_edge("Original", "Sasha", relationship = "Aunt #1")
G.add_edge("Original", "Owl", relationship = "Uncle")
G.add_edge("Original", "Thea", relationship = "Adopted child")
G.add_edge("Alexander", "Original", relationship = "Uncle")
G.add_edge("Alexander", "Thea", relationship = "Flirting")
G.add_edge("Alexander", "Owl", relationship = "Flirting")
G.add_edge("Alexander", "Sasha", relationship = "Dating")
G.add_edge("Alexander", "Diana", relationship = "Dating")


# Draw the graph
pos = nx.spring_layout(G)  # Define the layout for the graph
nx.draw(G, pos, with_labels=True, node_size=700, node_color='skyblue', font_size=10, font_weight='bold', edge_color='gray')
edge_labels = nx.get_edge_attributes(G, 'relationship')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

# Save the graph to an image file
plt.savefig("identity_system_graph.png", format="PNG")
plt.show()
```

```python
connections_per_node = {node: G.degree(node) for node in G.nodes()}

# Print the connections per node
for node, connections in connections_per_node.items():
    print(f"{node}: {connections} connections")
```
