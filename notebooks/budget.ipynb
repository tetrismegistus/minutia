```python
import matplotlib.pyplot as plt
import numpy as np

# Initial data and assumptions
initial_assets = 19.25
monthly_income = 0

monthly_expenses = {
    "rent": 0,
    "entertainment": 0,
    "water_gas": 0,
    "electric": 0,
    "internet + cellular": 58.92,
    "car": 635.24,
    "grocery": 0,
    "credit": 63,
    "daily": 0,
    "webhost": 10,
    "DigitalOcean": 7.2,
    "chatgpt": 20,
    "USAA": 228.42,
    "pet_food": 80,
    "gasoline": 150,
}


total_tax = 0
monthly_tax = 0

total_monthly_expenses = sum(monthly_expenses.values())
net_monthly_savings = monthly_income - total_monthly_expenses
months = np.arange(1, 13)
# Calculating assets over time
assets_over_time = [initial_assets + net_monthly_savings * month for month in months]

per_day_contract = (total_monthly_expenses - monthly_expenses['rent'] - initial_assets) / 30
print(f'total_monthly_expenses: {total_monthly_expenses}')
      
# Plotting the graph
plt.figure(figsize=(10, 6))
plt.plot(months, assets_over_time, marker='o', color='b', label='Assets Over Time')
plt.title('Financial Projection Over One Year')
plt.xlabel('Time (Months)')
plt.ylabel('Assets ($)')
plt.grid(True)
plt.xticks(months)
plt.legend()
plt.tight_layout()
plt.show()

# Creating the pie chart
plt.figure(figsize=(10, 8))
plt.pie(monthly_expenses.values(), labels=monthly_expenses.keys(), autopct='%1.1f%%', startangle=140)
plt.title('Monthly Expense Breakdown')
plt.show()
```
